```python
import requests
import justext
```

```python
headers = {"User-Agent": "Mozilla/5.0 (compatible; NLP-Study/1.0;)"}
response = requests.get("https://en.wikipedia.org/wiki/Toronto_Transit_Commission", headers=headers)
text = justext.justext(response.content, justext.get_stoplist("English"))

for content in text:
    if content.is_boilerplate:
        print(content.text)
```
